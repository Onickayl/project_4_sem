# 3D-графики 

Собрав погодные данные и параметры листа, можно строить различные 3D-графики.

Возможные параметры:
- день                                  (Day)
- температура                           (Temp)
- солнце                                (Sun)
- осадки                                (Rain)
- ветер                                 (Wind)
- среднее количество воды в листе        (AvgWater)
- среднее количество сахара в листе      (AvgSugar)
- среднее количество хлорофилла         (AvgChlorophyll)
- среднее количество антоциан            (AvgAnthocyanin)

Данные берутся из файлов "weather_.txt" и "leaf_.csv"

Графики загружаются в файлы "tree_.html".

```python
import pandas as pd
import plotly
import plotly.graph_objects as go
```

```python
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# загружаем данные
weather = pd.read_csv("weather_sim_default_normal.txt")
leaf = pd.read_csv("leaf_sim_default_normal.csv")

# объединяем две таблицы по столбцам Day и Season, чтобы всё синхронизировалось
merged = pd.merge(weather, leaf, on=['Day', 'Season'])

# 1 график: day, sun, temp
fig1 = go.Figure()

# добавляем 3D-линию
fig1.add_trace(go.Scatter3d(
    
    x=merged['Day'],
    y=merged['Sun'],
    z=merged['Temp'],

    mode='lines+markers',     # рисовать и линию между точками, и точки (маркеры)

    # настройка точек: размер 3 пикселя, цвет зависит от температуры (синий — холодно, красный — жарко)
    marker=dict(size=3, color=merged['Temp'], colorscale='Thermal', showscale=True),

    line=dict(color='orange', width=2),    # настройка линии: оранжевая, толщина 2

    name='Температура vs Солнце'    # имя легенды
))

# настройки осей
fig1.update_layout(
    title='Погода: День, Солнце, Температура',
    scene=dict(
        xaxis_title='День',
        yaxis_title='Солнце (%)',
        zaxis_title='Температура (°C)',
        aspectmode='data'               # подгоняет под данные
    )
)


# 2 график: day, water, cholor
fig2 = go.Figure()

# добавляем 3D-линию
fig2.add_trace(go.Scatter3d(
    
    x=merged['Day'],
    y=merged['AvgWater'],
    z=merged['AvgChlorophyll'],

    mode='lines+markers',     # рисовать и линию между точками, и точки (маркеры)

    # настройка точек: размер 3 пикселя, цвет зависит от % хлорофилла (синий — мало, красный — много)
    marker=dict(size=3, color=merged['AvgChlorophyll'], colorscale='Thermal', showscale=True),

    line=dict(color='green', width=2),    # настройка линии: зелёная, толщина 2

    name='Хлорофилл vs Вода'    # имя легенды
))

# настройки осей
fig2.update_layout(
    title='Листья: День, Вода, Хлорофилл',
    scene=dict(
        xaxis_title='День',
        yaxis_title='Вода в листьях (%)',
        zaxis_title='Хлорофилл (%)',
        aspectmode='data'               # подгоняет под данные
    )
)


plotly.offline.plot(fig1, filename='tree_weather.html', auto_open=False)
plotly.offline.plot(fig2, filename='tree_leaf.html', auto_open=False)
```
