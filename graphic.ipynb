Графики (пока что 2D)
- температуры (красный) и хлорофилла (зелёный) от дня
- солнца от дня
- дождя от дня
- ветра от дня

Данные берутся из файла data_of_simulation.csv

Картинка загружается в файл tree_simulation.png

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python

# читаем CSV
df = pd.read_csv('data_of_simulation.csv')
```

```python
# создаём график
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle('Tree Simulation - One Year', fontsize=14)

# 1 график - температура и хлорофилл
ax1 = axes[0, 0]
ax1.plot(df['Day'], df['Temp'], 'r-', label='Temperature (°C)', linewidth=1.5)
ax1.set_ylabel('Temperature (°C)', color='r')
ax1.tick_params(axis='y', labelcolor='r')

ax1b = ax1.twinx()
ax1b.plot(df['Day'], df['AvgChlorophyll'], 'g-', label='Chlorophyll (%)', linewidth=1.5)
ax1b.set_ylabel('Chlorophyll (%)', color='g')
ax1b.tick_params(axis='y', labelcolor='g')
ax1.set_title('Temperature & Chlorophyll')

# 2 график - солнце
ax2 = axes[0, 1]
ax2.plot(df['Day'], df['Sun'], 'y-', linewidth=1.5)
ax2.fill_between(df['Day'], 0, df['Sun'], alpha=0.2, color='yellow')
ax2.set_title('Sun (%)')
ax2.set_ylabel('Sun (%)')
ax2.set_ylim(0, 100)

# 3 график - дождь
ax3 = axes[1, 0]
ax3.plot(df['Day'], df['Rain'], 'b-', linewidth=1.5)
ax3.fill_between(df['Day'], 0, df['Rain'], alpha=0.2, color='blue')
ax3.set_title('Rain (%)')
ax3.set_ylabel('Rain (%)')
ax3.set_ylim(0, 100)

# 4 график - ветер
ax4 = axes[1, 1]
ax4.plot(df['Day'], df['Wind'], 'purple', linewidth=1.5)
ax4.fill_between(df['Day'], 0, df['Wind'], alpha=0.2, color='purple')
ax4.set_title('Wind (%)')
ax4.set_ylabel('Wind (%)')
ax4.set_ylim(0, 100)

# метки сезонов на все графики
for ax in [axes[0,0], axes[0,1], axes[1,0], axes[1,1]]:
    ax.set_xlabel('Day')
    ax.axvspan(0, 91, alpha=0.1, color='green', label='Spring')
    ax.axvspan(91, 182, alpha=0.1, color='yellow', label='Summer')
    ax.axvspan(182, 273, alpha=0.1, color='orange', label='Autumn')
    ax.axvspan(273, 365, alpha=0.1, color='lightblue', label='Winter')
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('tree_simulation.png', dpi=150, bbox_inches='tight')
plt.show()

print("Graph saved as tree_simulation.png")
```
